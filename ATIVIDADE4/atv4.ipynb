```python
import pandas as pd
```

```python
tabela5 = pd.read_excel('CD2022_Populacao_2010_Compatibilizada_20231222.xlsx', header=2)
tabela5 = tabela5.drop(columns=['Unnamed: 0'])
tabela5.columns = ['UF','COD_UF','COD_MUNIC','NOME_MUNIC','POP_2010_sinopse','POP_2010_compat','POP_2022']
ufs = ['RO','AC','AM','RR','PA','AP','TO','MA','PI','CE','RN','PB','PE','AL','SE','BA','MG','ES','RJ','SP','PR','SC','RS','MS','MT','GO','DF']
tabela5 = tabela5[tabela5['UF'].isin(ufs)].copy()
tabela5['POP_2010_compat'] = pd.to_numeric(tabela5['POP_2010_compat'], errors='coerce')
tabela5['POP_2022'] = pd.to_numeric(tabela5['POP_2022'], errors='coerce')
tabela5.head()
```

```python
populacao_por_estado = (
    tabela5.groupby('UF')[['POP_2010_compat','POP_2022']]
    .sum()
    .reset_index()
)
populacao_por_estado['CRESCIMENTO_2010_2022'] = populacao_por_estado['POP_2022'] - populacao_por_estado['POP_2010_compat']
populacao_por_estado = populacao_por_estado.sort_values('CRESCIMENTO_2010_2022', ascending=False)
populacao_por_estado
```

```python
populacao_por_estado.to_csv('populacao_por_estado.csv', index=False, encoding='utf-8-sig')
print('Arquivo populacao_por_estado.csv salvo com sucesso!')
```

```python
populacao_por_municipio = (
    tabela5.groupby(['UF','COD_UF','COD_MUNIC','NOME_MUNIC'])[['POP_2010_compat','POP_2022']]
    .sum()
    .reset_index()
)
populacao_por_municipio['CRESCIMENTO_2010_2022'] = populacao_por_municipio['POP_2022'] - populacao_por_municipio['POP_2010_compat']
populacao_por_municipio = populacao_por_municipio.sort_values('CRESCIMENTO_2010_2022', ascending=False)
populacao_por_municipio
```

```python
populacao_por_municipio.to_csv('populacao_por_municipio.csv', index=False, encoding='utf-8-sig')
print('Arquivo populacao_por_municipio.csv salvo com sucesso!')
```
